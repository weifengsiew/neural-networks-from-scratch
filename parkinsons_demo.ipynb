# Neural Network From Scratch: Parkinson's Dataset

This notebook applies the NumPy neural-network implementation in `neural_network.py` to a fixed train/test split of the Parkinson's disease dataset. Features are standardized using training data only, then the model is trained with mini-batch gradient descent.

```python
from pathlib import Path

import numpy as np

from neural_network import NeuralNetwork, compute_metrics_over_instances
```

```python
data_dir = Path('data')
train = np.genfromtxt(data_dir / 'parkinsons_train.csv', delimiter=',', skip_header=1)
test = np.genfromtxt(data_dir / 'parkinsons_test.csv', delimiter=',', skip_header=1)

X_train, y_train = train[:, :-1], train[:, -1].astype(int)
X_test, y_test = test[:, :-1], test[:, -1].astype(int)

feature_mean = X_train.mean(axis=0)
feature_scale = X_train.std(axis=0)
feature_scale[feature_scale == 0] = 1.0
X_train = (X_train - feature_mean) / feature_scale
X_test = (X_test - feature_mean) / feature_scale

print(f'train shape: {X_train.shape}')
print(f'test shape: {X_test.shape}')
```

```python
network = NeuralNetwork(num_hidden_layers=1, num_neurons_per_hidden_layer=8)
network.configure_for_fit(regularization_strength=0.01, step_size=0.1)

history = network.fit(
    X_train,
    y_train,
    num_iterations=100,
    batch_size=32,
    random_seed=7,
    shuffle=True,
    record_history=True,
    X_test=X_test,
    y_test=y_test,
)

test_accuracy, test_f1 = compute_metrics_over_instances(network, X_test, y_test)
print(f'test accuracy: {test_accuracy:.3f}')
print(f'test F1 score: {test_f1:.3f}')
```

```python
train_accuracy, train_f1 = compute_metrics_over_instances(network, X_train, y_train)
print(f'final train accuracy: {train_accuracy:.3f}')
print(f'final train F1 score: {train_f1:.3f}')
print(f'weight updates recorded: {len(history["updates"])}')
```
